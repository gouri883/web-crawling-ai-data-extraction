```python
from crawl4ai.deep_crawling import BFSDeepCrawlStrategy
from crawl4ai.deep_crawling import FilterChain, SEOFilter
```

```python
from crawl4ai import AsyncWebCrawler
from crawl4ai.async_configs import BrowserConfig, CrawlerRunConfig
from crawl4ai.deep_crawling import BFSDeepCrawlStrategy, FilterChain, SEOFilter

browser_config = BrowserConfig(
    headless=True
)

seo_filter = SEOFilter(
    threshold=0.5,
    keywords=["tutorial", "guide", "documentation"]
)

deep_strategy = BFSDeepCrawlStrategy(
    max_depth=1,
     max_pages=3,
    filter_chain=FilterChain([seo_filter])
)

run_config = CrawlerRunConfig(
    deep_crawl_strategy=deep_strategy
)

async def main():
    async with AsyncWebCrawler(config=browser_config) as crawler:
        results = await crawler.arun(
            url="https://www.bbc.com/news",
            config=run_config
        )

        print(f"✅ Pages crawled: {len(results)}\n")

        for i, page in enumerate(results[:5], 1):
            print(f"--- Page {i} ---")
            print(page.url)
            print(page.markdown[:20])
            print()

await main()

```

```python
import asyncio
from crawl4ai import AsyncWebCrawler
from crawl4ai.async_configs import BrowserConfig, CrawlerRunConfig

async def main():
    run_config = CrawlerRunConfig(
        verbose=True,
        page_timeout=15000   # wait up to 15 seconds
    )

    browser_config = BrowserConfig(
        headless=True
    )

    async with AsyncWebCrawler(config=browser_config) as crawler:
        result = await crawler.arun(
            url="https://www.bbc.com/news",
            config=run_config
        )

        if result.success:
            images = result.media.get("images", [])
            print("✅ Images found:", len(images))

            for i, img in enumerate(images[:5], 1):
                print(f"[Image {i}] {img.get('src')}")
        else:
            print("❌ Error:", result.error_message)

await main()
```

```python
import asyncio
from crawl4ai import AsyncWebCrawler
from crawl4ai.async_configs import BrowserConfig, CrawlerRunConfig
from bs4 import BeautifulSoup
from collections import Counter
import re
```

```python
async def crawl_bbc():
    run_config = CrawlerRunConfig(
        page_timeout=15000
    )
    browser_config = BrowserConfig(headless=True)

    async with AsyncWebCrawler(config=browser_config) as crawler:
        result = await crawler.arun(
            url="https://www.bbc.com/news",
            config=run_config
        )
        return result.cleaned_html, result.media
```

```python
# Run this in ONE cell

from bs4 import BeautifulSoup

html, media = await crawl_bbc()
soup = BeautifulSoup(html, "html.parser")

print("HTML length:", len(html))
print("Images found:", len(media.get("images", [])))

```

Extract All BBC News Headlines

```python
headlines = set()

for tag in soup.find_all(["h1", "h2", "h3"]):
    text = tag.get_text(strip=True)
    if len(text) > 30:
        headlines.add(text)

print("📰 Headlines:")
for h in list(headlines)[:10]:
    print("-", h)
```

Category-wise News Link Extraction

```python
categories = {
    "World": [],
    "Business": [],
    "Technology": [],
    "Sport": [],
    "Health": []
}

# Collect links
for a in soup.find_all("a", href=True):
    href = a["href"]

    for cat in categories:
        if f"/{cat.lower()}" in href:
            # Convert relative URLs to full URLs
            if href.startswith("/"):
                href = "https://www.bbc.com" + href

            categories[cat].append(href)

# Print results
for cat, links in categories.items():
    unique_links = sorted(set(links))

    print(f"\n🟢 {cat} ({len(unique_links)} links)")
    for link in unique_links[:10]:   # print first 10 to avoid clutter
        print("  -", link)

```

Homepage vs technology page comparison

```python
async def crawl_page(url):
    async with AsyncWebCrawler(config=BrowserConfig(headless=True)) as crawler:
        result = await crawler.arun(url)
        return result.cleaned_html

home_html = await crawl_page("https://www.bbc.com/news")
tech_html = await crawl_page("https://www.bbc.com/news/technology")

print("Homepage length:", len(home_html))
print("Technology page length:", len(tech_html))
```

Image to text ratio analysis

```python
text = soup.get_text(separator=" ")
word_count = len(text.split())
image_count = len(media.get("images", []))

print("Words:", word_count)
print("Images:", image_count)
print("Image-to-text ratio:", round(image_count / max(word_count,1), 6))
```

Key word frequency analysis

```python
keywords = ["war", "election", "economy", "climate", "technology"]

words = re.findall(r"\b\w+\b", text.lower())
counter = Counter(words)

print("📊 Keyword Frequency:")
for k in keywords:
    print(k, "→", counter[k])
```

Internal vs extrenal link detection

```python
internal = 0
external = 0

for a in soup.find_all("a", href=True):
    href = a["href"]
    if href.startswith("https://www.bbc.com") or href.startswith("/"):
        internal += 1
    elif href.startswith("http"):
        external += 1

print("🔗 Internal links:", internal)
print("🌍 External links:", external)
```

Sentiment trend analysis

```python
from crawl4ai import AsyncWebCrawler
from crawl4ai.async_configs import BrowserConfig, CrawlerRunConfig
from bs4 import BeautifulSoup

async def crawl_bbc():
    async with AsyncWebCrawler(config=BrowserConfig(headless=True)) as crawler:
        result = await crawler.arun(
            url="https://www.bbc.com/news",
            config=CrawlerRunConfig(page_timeout=15000)
        )
        return result.cleaned_html

html = await crawl_bbc()
soup = BeautifulSoup(html, "html.parser")
```

```python
headlines = []

for tag in soup.find_all(["h1", "h2", "h3"]):
    text = tag.get_text(strip=True)
    if 30 < len(text) < 120:
        headlines.append(text)

headlines = list(set(headlines))
print("Total headlines:", len(headlines))
```

```python
from nltk.sentiment.vader import SentimentIntensityAnalyzer
import nltk

nltk.download("vader_lexicon")

sia = SentimentIntensityAnalyzer()
```

“BBC News headlines were crawled and sentiment analysis was applied using VADER to study the emotional tone of current news.”

```python
sentiment_map = {
    "positive": [],
    "negative": [],
    "neutral": []
}

for headline in headlines:
    score = sia.polarity_scores(headline)["compound"]

    if score >= 0.05:
        sentiment_map["positive"].append((headline, score))
    elif score <= -0.05:
        sentiment_map["negative"].append((headline, score))
    else:
        sentiment_map["neutral"].append((headline, score))
```

```python
for sentiment, items in sentiment_map.items():
    print(f"\n * {sentiment.upper()} HEADLINES ({len(items)})")

    for text, score in items[:5]:   # print first 5 to keep output clean
        print(f"  - {text}  [score={score:.2f}]")
```
